```python
import pandas as pd
```

```python
orders = pd.read_csv(r"C:\Users\w7127363\Documents\olist_orders_dataset.csv")
reviews = pd.read_csv(r"C:\Users\w7127363\Documents\olist_order_reviews_dataset.csv")
order_items = pd.read_csv(r"C:\Users\w7127363\Documents\olist_order_items_dataset.csv")
products = pd.read_csv(r"C:\Users\w7127363\Documents\olist_products_dataset.csv")
customers = pd.read_csv(r"C:\Users\w7127363\Documents\olist_customers_dataset.csv")
sellers = pd.read_csv(r"C:\Users\w7127363\Documents\olist_sellers_dataset.csv")
payments = pd.read_csv(r"C:\Users\w7127363\Documents\olist_order_payments_dataset.csv")
```

```python
print("Orders:", orders.shape)
print("Reviews:", reviews.shape)
print("Order Items:", order_items.shape)
print("Products:", products.shape)
print("Customers:", customers.shape)
print("Sellers:", sellers.shape)
print("Payments:", payments.shape)
```

```python
print("Orders columns:")
print(orders.columns.tolist())

print("\nReviews columns:")
print(reviews.columns.tolist())

print("\nOrder Items columns:")
print(order_items.columns.tolist())

print("\nProducts columns:")
print(products.columns.tolist())

print("\nCustomers columns:")
print(customers.columns.tolist())

print("\nSellers columns:")
print(sellers.columns.tolist())

print("\nPayments columns:")
print(payments.columns.tolist())
```

```python
print("Missing values in Orders:")
print(orders.isnull().sum())

print("\nMissing values in Reviews:")
print(reviews.isnull().sum())

print("\nMissing values in Order Items:")
print(order_items.isnull().sum())

print("\nMissing values in Products:")
print(products.isnull().sum())

print("\nMissing values in Customers:")
print(customers.isnull().sum())

print("\nMissing values in Sellers:")
print(sellers.isnull().sum())

print("\nMissing values in Payments:")
print(payments.isnull().sum())
```

```python
print("Duplicate rows:")

print("Orders:", orders.duplicated().sum())
print("Reviews:", reviews.duplicated().sum())
print("Order Items:", order_items.duplicated().sum())
print("Products:", products.duplicated().sum())
print("Customers:", customers.duplicated().sum())
print("Sellers:", sellers.duplicated().sum())
print("Payments:", payments.duplicated().sum())
```

```python
date_columns_orders = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns_orders:
    orders[column] = pd.to_datetime(orders[column], errors="coerce")

reviews["review_creation_date"] = pd.to_datetime(
    reviews["review_creation_date"],
    errors="coerce"
)

reviews["review_answer_timestamp"] = pd.to_datetime(
    reviews["review_answer_timestamp"],
    errors="coerce"
)

order_items["shipping_limit_date"] = pd.to_datetime(
    order_items["shipping_limit_date"],
    errors="coerce"
)
```

```python
print(orders[date_columns_orders].dtypes)

print("\nReview date types:")
print(reviews[["review_creation_date", "review_answer_timestamp"]].dtypes)

print("\nShipping date type:")
print(order_items["shipping_limit_date"].dtype)
```

```python
reviews["target"] = reviews["review_score"].apply(
    lambda score: 0 if score <= 2 else 1
)
```

```python
print("Target distribution:")
print(reviews["target"].value_counts())

print("\nTarget percentages:")
print(reviews["target"].value_counts(normalize=True) * 100)
```

```python
orders_reviews = orders.merge(
    reviews[["order_id", "review_score", "target"]],
    on="order_id",
    how="inner"
)

print("Merged dataset shape:")
print(orders_reviews.shape)

print("\nColumns:")
print(orders_reviews.columns.tolist())
```

```python
order_items_summary = order_items.groupby("order_id").agg(
    number_of_items=("order_item_id", "count"),
    total_price=("price", "sum"),
    total_freight=("freight_value", "sum")
).reset_index()

print("Order items summary:")
print(order_items_summary.head())

print("\nShape:")
print(order_items_summary.shape)
```

```python
orders_reviews = orders_reviews.merge(
    order_items_summary,
    on="order_id",
    how="left"
)

print("Dataset shape after adding order item information:")
print(orders_reviews.shape)

print("\nColumns:")
print(orders_reviews.columns.tolist())
```

```python
order_product = order_items.merge(
    products,
    on="product_id",
    how="left"
)

print("Order product shape:")
print(order_product.shape)

print("\nColumns:")
print(order_product.columns.tolist())
```

```python
product_summary = order_product.groupby("order_id").agg(
    number_of_products=("product_id", "nunique"),
    average_product_weight=("product_weight_g", "mean"),
    average_product_length=("product_length_cm", "mean"),
    average_product_height=("product_height_cm", "mean"),
    average_product_width=("product_width_cm", "mean"),
    average_product_photos=("product_photos_qty", "mean")
).reset_index()

print("Product summary:")
print(product_summary.head())

print("\nShape:")
print(product_summary.shape)
```

```python
orders_reviews = orders_reviews.merge(
    product_summary,
    on="order_id",
    how="left"
)

print("Dataset shape after adding product information:")
print(orders_reviews.shape)

print("\nColumns:")
print(orders_reviews.columns.tolist())
```

```python
payments_summary = payments.groupby("order_id").agg(
    total_payment_value=("payment_value", "sum"),
    payment_installments=("payment_installments", "max"),
    number_of_payments=("payment_sequential", "count")
).reset_index()

print("Payment summary:")
print(payments_summary.head())

print("\nShape:")
print(payments_summary.shape)
```

```python
orders_reviews = orders_reviews.merge(
    payments_summary,
    on="order_id",
    how="left"
)

print("Dataset shape after adding payment information:")
print(orders_reviews.shape)

print("\nColumns:")
print(orders_reviews.columns.tolist())
```

```python
orders_reviews = orders_reviews.merge(
    customers,
    on="customer_id",
    how="left"
)

print("Dataset shape after adding customer information:")
print(orders_reviews.shape)

print("\nColumns:")
print(orders_reviews.columns.tolist())
```

```python
order_seller = order_items.merge(
    sellers,
    on="seller_id",
    how="left"
)

print("Order seller shape:")
print(order_seller.shape)

print("\nColumns:")
print(order_seller.columns.tolist())
```

```python
seller_summary = order_seller.groupby("order_id").agg(
    number_of_sellers=("seller_id", "nunique"),
    seller_zip_code=("seller_zip_code_prefix", "first"),
    seller_city=("seller_city", "first"),
    seller_state=("seller_state", "first")
).reset_index()

print("Seller summary:")
print(seller_summary.head())

print("\nShape:")
print(seller_summary.shape)
```

```python
orders_reviews = orders_reviews.merge(
    seller_summary,
    on="order_id",
    how="left"
)

print("Dataset shape after adding seller information:")
print(orders_reviews.shape)

print("\nColumns:")
print(orders_reviews.columns.tolist())
```

```python
orders_reviews["delivery_days"] = (
    orders_reviews["order_delivered_customer_date"]
    - orders_reviews["order_purchase_timestamp"]
).dt.days

orders_reviews["delivery_delay_days"] = (
    orders_reviews["order_delivered_customer_date"]
    - orders_reviews["order_estimated_delivery_date"]
).dt.days

orders_reviews["delivery_missing"] = (
    orders_reviews["order_delivered_customer_date"].isna().astype(int)
)

print("Delivery features created:")
print(
    orders_reviews[
        ["delivery_days", "delivery_delay_days", "delivery_missing"]
    ].head()
)

print("\nMissing values:")
print(
    orders_reviews[
        ["delivery_days", "delivery_delay_days"]
    ].isna().sum()
)
```

```python
missing_values = orders_reviews.isna().sum()

missing_values = missing_values[missing_values > 0].sort_values(
    ascending=False
)

print("Missing values in the processed dataset:")
print(missing_values)
```

```python
# Fill numerical feature missing values with the median
numeric_columns = [
    "average_product_photos",
    "average_product_length",
    "average_product_height",
    "average_product_width",
    "average_product_weight",
    "number_of_products",
    "total_freight",
    "total_price",
    "number_of_sellers",
    "seller_zip_code",
    "number_of_items",
    "total_payment_value",
    "payment_installments",
    "number_of_payments"
]

for column in numeric_columns:
    orders_reviews[column] = orders_reviews[column].fillna(
        orders_reviews[column].median()
    )

# Fill categorical missing values
categorical_columns = [
    "seller_city",
    "seller_state"
]

for column in categorical_columns:
    orders_reviews[column] = orders_reviews[column].fillna("Unknown")

# Fill missing approval dates with the earliest available date
orders_reviews["order_approved_at"] = orders_reviews[
    "order_approved_at"
].fillna(orders_reviews["order_purchase_timestamp"])

print("Missing values after handling:")
print(orders_reviews.isna().sum()[orders_reviews.isna().sum() > 0])
```

```python
columns_to_remove = [
    "review_score",
    "order_id",
    "customer_id",
    "customer_unique_id",
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

orders_reviews = orders_reviews.drop(
    columns=columns_to_remove
)

print("Final preprocessing dataset shape:")
print(orders_reviews.shape)

print("\nRemaining columns:")
print(orders_reviews.columns.tolist())
```

```python
print("Final dataset shape:")
print(orders_reviews.shape)

print("\nDuplicate rows:")
print(orders_reviews.duplicated().sum())

print("\nTarget distribution:")
print(orders_reviews["target"].value_counts())

print("\nTarget proportions:")
print(orders_reviews["target"].value_counts(normalize=True))

print("\nData types:")
print(orders_reviews.dtypes)
```
