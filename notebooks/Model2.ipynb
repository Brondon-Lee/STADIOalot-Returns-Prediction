```python
import pandas as pd
import scipy.sparse

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# Load feature-engineered data
X_train = scipy.sparse.load_npz(
    r"C:\Users\w7127363\Downloads\feature_engineered\X_train.npz"
)

X_test = scipy.sparse.load_npz(
    r"C:\Users\w7127363\Downloads\feature_engineered\X_test.npz"
)

y_train = pd.read_csv(
    r"C:\Users\w7127363\Downloads\feature_engineered\y_train.csv"
).squeeze()

y_test = pd.read_csv(
    r"C:\Users\w7127363\Downloads\feature_engineered\y_test.csv"
).squeeze()

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
model = RandomForestClassifier(
    n_estimators=200,
    max_depth=15,
    min_samples_split=5,
    random_state=42,
    n_jobs=-1
)

print("Random Forest model created.")
print(model)
```

```python
print("Training Random Forest model...")

model.fit(X_train, y_train)

print("Training completed successfully.")
```

```python
y_pred = model.predict(X_test)
y_pred_proba = model.predict_proba(X_test)[:, 1]

print("Predictions completed.")
print("Number of predictions:", len(y_pred))
```

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_pred_proba)

print("Random Forest Results")
print("---------------------")
print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-score:  {f1:.4f}")
print(f"ROC-AUC:   {roc_auc:.4f}")
```

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix")
print("----------------")
print(cm)
```

```python
print("Classification Report")
print("---------------------")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=["Negative Outcome", "Positive Outcome"]
    )
)
```

```python
results = pd.DataFrame({
    "Model": ["Random Forest"],
    "Accuracy": [accuracy],
    "Precision": [precision],
    "Recall": [recall],
    "F1_Score": [f1],
    "ROC_AUC": [roc_auc]
})

results_path = r"C:\Users\w7127363\Downloads\model2_results.csv"

results.to_csv(results_path, index=False)

print("Random Forest results saved successfully.")
print(results)
print("\nSaved to:")
print(results_path)
```

```python
import joblib

model_path = r"C:\Users\w7127363\Downloads\random_forest_model.pkl"

joblib.dump(model, model_path)

print("Random Forest model saved successfully.")
print(model_path)
```
